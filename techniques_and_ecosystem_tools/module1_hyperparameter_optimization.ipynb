* Optimizaton - finding the optimal value

* Goal to maximize accuracy
* Accuracy - sometimes misleading - when dataset imbalanced
* Precision - TP/(TP+FP). Use when false positives are costly.
* Recall - TP/(TP+FN). Use when false negatives  are costly.
* F1 = harmonic mean of Pr and Re. It is preferred over accuracy with imbalanced datasets
* Torchmetrics has builtin classes for the metrics. can also calculate macro or micro averaging.




# Typical Hyperparameters:

* Effect of learning rate on accuracy : Accuracy vs Learning Rate
* Maximize: Accuracy, Precision, Recall, F1 Score
* Minimize: Training/ Inference Time, Mean squared error, log loss
* External factors affecting performance - data quantity, data quality, feature representation 
* Internal: number of layers, dropout, learning rate, batch size

## Learning Rate Scheduler

* Start with high learning rate - rapidly increases validation accuracy in the initial epochs
* gradually decrease the learning rate to finetune the model - this could get the higher val accuracy in the least number of epochs hence reducing the training time.
* PyTorch builtin LR schedulers - StepLR, ReduceLROnPlateau, CosineAnnealingLR
* StepLR - Gamma (Factor of rate decay), StepSize (number of epochs) - when model benefits from periodic substantial reductions in learning rate

`scheduler = optim.lr_scheduler.StepLR(optimizer, step_size = 10, gamma = 0.2)`

`...`

`scheduler.step()` <- at the end of each epoch 

* ReduceLROnPlateau - Factor(Factor of rate decay), Patience (Number of epochs without improvement) - reduces the learning rate when the model's performance stalls

`scheduler = optim.lr_scheduler.ReduceLROnPlateauLR(optimizer, mode = 'max', factor = 0.2, patience = 3) `

`...`

`scheduler.step(val_acc) <- decide whether lr should be reduced based on val_acc`


* Cosine Annealing - T_max (max number of iterations), eta_min (min learning rate) - gradually lowers the learning rate following a cosine curve from its initial value to a minimum value over a predetermined number of epochs.

* Provides a smoother transition than the other two schedulers

`scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer,T_max = n_epochs, eta_min = 0.0002) `

`...`

`scheduler.step()`
